## Q1. Arithmetic and Assignment Operators

```python
a = 20
b = 6

print("a =", a)
print("b =", b)

print("\n-- Arithmetic Operators --")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

print("\n-- Assignment Operators --")
a += 5
print("After a += 5:", a)
a -= 3
print("After a -= 3:", a)
a *= 2
print("After a *= 2:", a)
a /= 4
print("After a /= 4:", a)
```

## Q2. Comparison and Logical Operators

```python
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print("\n-- Comparison Operators --")
print("x == y :", x == y)
print("x != y :", x != y)
print("x > y  :", x > y)
print("x < y  :", x < y)
print("x >= y :", x >= y)
print("x <= y :", x <= y)

print("\n-- Logical Operators --")
print("x > 0 and y > 0 :", x > 0 and y > 0)
print("x > 10 or y > 10 :", x > 10 or y > 10)
print("not (x == y) :", not (x == y))
```

## Q3. Bitwise Operators

```python
p = 12
q = 10

print("p =", p)
print("q =", q)

print("\n-- Bitwise Operators --")
print("AND  (p & q) :", p & q)
print("OR   (p | q) :", p | q)
print("XOR  (p ^ q) :", p ^ q)
print("NOT  (~p)    :", ~p)
print("Left Shift  (p << 2) :", p << 2)
print("Right Shift (p >> 2) :", p >> 2)
```

## Q4. Identity and Membership Operators

```python
numbers = [10, 20, 30, 40, 50]

var1 = numbers
var2 = [10, 20, 30, 40, 50]

print("List:", numbers)

print("\n-- Identity Operators --")
print("var1 is numbers     :", var1 is numbers)
print("var2 is not numbers :", var2 is not numbers)

print("\n-- Membership Operators --")
print("30 in numbers     :", 30 in numbers)
print("99 not in numbers :", 99 not in numbers)
```

## Q5. String Operations

```python
sentence = "Data Science is the future of technology"

print("Original:", sentence)
print("Length:", len(sentence))
print("Uppercase:", sentence.upper())
print("Lowercase:", sentence.lower())
print("'future' in string:", "future" in sentence)
print("After replace:", sentence.replace("future", "backbone"))
print("Sliced [0:12]:", sentence[0:12])
print("Reversed:", sentence[::-1])
```
